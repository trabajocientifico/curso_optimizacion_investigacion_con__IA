```python
import numpy as np
import pandas as pd
import plotly.express as px

# Simular 1000 datos para ventas y marketing
np.random.seed(42) # para reproducibilidad
marketing = np.random.rand(1000) * 100 # Valores de marketing entre 0 y 100
ventas = 50 + 2 * marketing + np.random.randn(1000) * 20 # Ventas con cierta dependencia de marketing y ruido

# Crear un DataFrame para Plotly Express
df = pd.DataFrame({'Gastos de Marketing': marketing, 'Ventas': ventas})

# Crear el gráfico de dispersión interactivo con Plotly Express
fig = px.scatter(df, x='Gastos de Marketing', y='Ventas',
                 title='<b>Gráfico de Dispersión de Ventas vs. Marketing</b>',
                 color_discrete_sequence=['blue'], # Puntos de color azul
                 opacity=0.7,
                 trendline='ols', # Añadir línea de tendencia OLS (Ordinary Least Squares)
                 trendline_color_override='red' # Color rojo para la línea de tendencia
                )

# Obtener la ecuación de la línea de tendencia desde el modelo OLS
# Nota: Plotly Express no expone directamente la ecuación en el mismo formato que Matplotlib.
# Se puede obtener de los resultados del modelo si es necesario, pero para este caso
# Plotly ya muestra el hover de la línea de tendencia con información relevante.

# Personalizar etiquetas (ya en negrita por el título principal)
fig.update_layout(
    xaxis_title='<b>Gastos de Marketing</b>',
    yaxis_title='<b>Ventas</b>',
    hovermode='x unified' # Mejora la interactividad del hover
)

# Mostrar la figura
fig.show()
```
